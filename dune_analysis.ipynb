```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import scipy.stats as stats
import statistics

df = pd.read_excel("dune_data.xlsx", sheet_name=None)
regression = df["REGRESSION USE"]
```

```python
sns.regplot(regression, y='Nitrogen Content', x="BEACHGRASS", ci=95, color="black")
slope, intercept, r_value, p_value, std_err = stats.linregress(regression["Nitrogen Content"], regression["BEACHGRASS"])
r_value**2, p_value, slope, intercept
plt.xlabel("Number of Beachgrass")
```

```python
t_stat, p_value = stats.ttest_rel(regression["Nitrogen Content"], regression["SHRUBS"])
print(t_stat, p_value)
slope, intercept, r_value, p_value, std_err = stats.linregress(regression["Nitrogen Content"], regression["SHRUBS"])
print(r_value**2)


for i in ['TREES','SHRUBS', 'WOODY', 'FORBS', 'BEACHGRASS', 'HERBACEOUS']:
    slope, intercept, r_value, p_value, std_err = stats.linregress(regression["Nitrogen Content"], regression[i])
    print(i, r_value**2, p_value)
```

```python
sns.regplot(regression, x='WOODY', y="Nitrogen Content", color="black", ci=95)
slope, intercept, r_value, p_value, std_err = stats.linregress(regression["WOODY"], regression["Nitrogen Content"])
print(r_value**2, p_value, slope, intercept)
plt.xlabel("Number of Woody Organisms")
```
